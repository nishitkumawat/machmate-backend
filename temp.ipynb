```python
import pandas as pd
import json
# Read dataset (update filename as needed)
df = pd.read_excel("District_Masters.xlsx")

# Build dictionary
states_and_districts = {}

for _, row in df.iterrows():
    state = row["State Name"].title().strip()
    district = row["District Name"].strip()

    if state not in states_and_districts:
        states_and_districts[state] = []
    
    states_and_districts[state].append(district)

# Save as JSON
with open("states_and_districts.json", "w", encoding="utf-8") as f:
    json.dump(states_and_districts, f, indent=2, ensure_ascii=False)

print("✅ JSON created in desired format!")
```
